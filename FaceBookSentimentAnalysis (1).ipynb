```python
import pandas as pd
import nltk
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
```

```python
nltk.downloader.download('vader_lexicon')
```

```python
file="/content/FacebookPosts.xlsx"
```

```python
xl = pd.ExcelFile(file)
```

```python
dfs=xl.parse(xl.sheet_names[0])
```

```python
dfs=list(dfs['Timeline'])
```

```python
print(dfs)
```

```python
sid=SentimentIntensityAnalyzer()
```

```python
for data in dfs:
  ss=sid.polarity_scores(data)
  print(data)
  for k in ss:
    print(k,ss[k])
```

```python
count = 0

for data in dfs:
    count += data.lower().count("tom")

print("Number of occurrences of Tom:", count)
```

```python
posts_with_tom = sum("tom" in data.lower() for data in dfs)

print("Number of posts containing Tom:", posts_with_tom)
```
